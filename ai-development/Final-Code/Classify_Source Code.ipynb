```python
import os
from ultralytics import YOLO

# Load a model
model = YOLO("yolov8m.pt")

# Train the model with W&B disabled
train_results = model.train(data="data.yaml", epochs=100, imgsz=640)
```

```python
result = model.predict("/content/Screenshot 2024-10-24 225233.png")
result[0].show()
```

```python
# Evaluate model performance on the validation set
metrics = model.val()
print(metrics)
```
